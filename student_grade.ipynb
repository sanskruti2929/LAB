```python
import pandas as pd

# Step 1: Read the CSV file
df = pd.read_csv('grade.csv')

# Step 2: Define the map function
def mapper(row):
    name = row['Student Name']
    marks = row['Marks']
    if marks >= 90:
        grade = 'A'
    elif marks >= 80:
        grade = 'B'
    elif marks >= 70:
        grade = 'C'
    elif marks >= 60:
        grade = 'D'
    else:
        grade = 'F'
    return (name, grade)

# Step 3: Apply the mapper to each row
mapped_results = df.apply(mapper, axis=1).tolist()

# Step 4: Define a simple reducer (no aggregation needed)
def reducer(mapped_data):
    return dict(mapped_data)

# Step 5: Get the final results
final_results = reducer(mapped_results)

# Step 6: Print the output
for student, grade in final_results.items():
    print(f"{student}: {grade}")
```
