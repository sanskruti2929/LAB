```python
from collections import defaultdict

# Step 1: Load the data from the text file
with open('matrix_multiplication.txt', 'r') as f:
    lines = f.readlines()

# Step 2: Simulate the mapper
def mapper(lines):
    intermediate = []
    num_rows_A = 2
    num_cols_B = 2

    for line in lines:
        matrix, i, j, value = line.strip().split()
        i, j, value = int(i), int(j), int(value)

        if matrix == "A":
            for k in range(num_cols_B):  
                intermediate.append(((i, k), ("A", j, value)))
        elif matrix == "B":
            for k in range(num_rows_A):  
                intermediate.append(((k, j), ("B", i, value)))

    return intermediate

# Step 3: Shuffle and sort (group by key)
def shuffle_and_sort(mapped_data):
    grouped = defaultdict(list)
    for key, value in mapped_data:
        grouped[key].append(value)
    return grouped

# Step 4: Simulate the reducer
def reducer(grouped_data):
    final_results = {}
    for key, values in grouped_data.items():
        A_vals = defaultdict(int)
        B_vals = defaultdict(int)

        for mat, index, val in values:
            if mat == "A":
                A_vals[index] = val
            else:
                B_vals[index] = val

        result = sum(A_vals[k] * B_vals[k] for k in A_vals if k in B_vals)
        final_results[key] = result

    return final_results

# Run the MapReduce steps
mapped = mapper(lines)
grouped = shuffle_and_sort(mapped)
final_result = reducer(grouped)

# Step 5: Display the results
print("Resultant Matrix (as key: value pairs):")
for key in sorted(final_result):
    print(f"{key}: {final_result[key]}")
```
